```python
import pandas as pd

df_dipendenti = pd.read_csv("dipendenti.csv")
df_dipartimenti = pd.read_csv("dipartimenti.csv")

df_dipendenti_copia = df_dipendenti.copy()
df_dipartimenti_copia = df_dipartimenti.copy()

df_dipe_and_dipa = pd.merge(
    df_dipendenti_copia,
    df_dipartimenti_copia,
    left_on="id_dipartimento",
    right_on="id_dipartimento",
    how="left"
)

riepilogo = (df_dipe_and_dipa.groupby("reparto", dropna=False).agg(
    totale_dipendenti = ("id_dipendente", "count"),
    stipendio_basso = ("stipendio", "min"),
    stipendio_alto = ("stipendio", "max")
)).reset_index()


print("=== Distribuzione stipendi ===")
print(df_dipendenti_copia["stipendio"].describe().round(2))

print("=== Calcolo Mediana ===")
print(f"Mediana: ", df_dipendenti_copia["stipendio"].median())

print("=== Stipendio Minimo e Massimo ===")
print(f"Stipendio Minimo: ", df_dipe_and_dipa.loc[df_dipe_and_dipa["stipendio"].idxmin()])
print(f"Stipendio Massimo: ", df_dipe_and_dipa.loc[df_dipe_and_dipa["stipendio"].idxmax()])

print("=== Riepilogo ===")
print(riepilogo)
```
