```python
from project_work import connect_db
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

engine = connect_db()

df_anzianita_analitica = pd.read_sql("SELECT * FROM analisi_anzianita", engine)
df_reparto = (
    df_anzianita_analitica.groupby("reparto", as_index=False)["media_stipendi"].first()
)

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.bar(
    df_reparto["reparto"],
    df_reparto["media_stipendi"]
)
ax.set_title("Stipendi Medi per Reparto")
ax.set_xlabel("Reparto")
ax.set_ylabel("Stipendio Medio")

plt.savefig("stipendi_medi_reparto_barchart.png", dpi=150)

plt.show()
```
